This notebook uses the project's reproducible Python model code. Choose `responded` or `converted` as the outcome, then run the cell. The data is synthetic; results are for demonstration only.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "model.py").exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))
from model import run_model

target = "responded"  # Change to "converted" to predict conversion
report = run_model(target)

print("Target:", report["prediction_target"])
print("Training through:", report["train_through_date"])
print("Testing from:", report["test_from_date"])
print("Held-out metrics:")
for metric, value in report["test_metrics"].items():
    print(f"  {metric}: {value}")
print("Baseline accuracy:", round(report["baseline_test_accuracy"], 3))
print("Top feature associations:")
for item in report["top_model_drivers"][:5]:
    print(f"  {item['feature']}: {item['direction']} ({item['coefficient']:+.4f})")
```

## Interpretation

The model uses a chronological holdout and selects its decision threshold using training data. Coefficients describe associations in this fitted model; they do not establish causes.
